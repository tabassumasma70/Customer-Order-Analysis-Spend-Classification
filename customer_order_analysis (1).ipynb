```python
                                             ****Task 1: Store Customer Orders****
Data Structuring
```

```python
List of customers
```

```python

customers = ["Alice", "Bob", "Charlie", "Diana"]

```

```python
 List of tuples: (customer_name, product, price, category)
```

```python

orders = [
    ("Alice", "Laptop", 1200.0, "Electronics"),
    ("Alice", "T-Shirt", 25.0, "Clothing"),
    ("Bob", "Headphones", 80.0, "Electronics"),
    ("Bob", "Coffee Maker", 60.0, "Home Essentials"),
    ("Charlie", "Jeans", 45.0, "Clothing"),
    ("Charlie", "Blender", 35.0, "Home Essentials"),
    ("Diana", "Smartphone", 800.0, "Electronics"),
    ("Diana", "Jacket", 110.0, "Clothing")
]
```

```python
Dictionary mapping customer names to lists of ordered products
```

```python
customer_orders = {}
for name, product, price, category in orders:
    customer_orders.setdefault(name, []).append(product)

print("Customer Orders Mapping:", customer_orders)
```

```python
                              ****Task 2: Classify Products by Category****
Dictionaries & Sets
Extract unique categories and product mappings:
```

```python
maps product to category using a dictionary
```

```python

product_catalog = {product: category for name, product, price, category in orders}

```

```python
Unique categories using a set
```

```python

categories = {category for name, product, price, category in orders}

print("Product Catalog:", product_catalog)
print("Available Categories:", categories)
```

```python
                  ****Task 3: Analyze Customer Orders & Spend Classification****
Loops & Conditionals
Calculate total spend per customer and classify them based on the spending thresholds 
```

```python
customer_totals = {}
customer_classes = {}
```

```python
# Calculate total spending per customer
```

```python
for name, product, price, category in orders:
    customer_totals[name] = customer_totals.get(name, 0.0) + price
```

```python
# Classify buyers
```

```python
for name, total in customer_totals.items():
    if total > 500:
        customer_classes[name] = "High-Value Buyer"
    elif 50 <= total <= 100:
        customer_classes[name] = "Low-Value Buyer"
    else:
        customer_classes[name] = "Moderate-Value Buyer"

print("Total Spend:", customer_totals)
print("Classifications:", customer_classes)
```

```python
                            ****Task 4: Generate Business Insights****
Aggregations & Comprehensions
Calculate category revenue, retrieve unique products, filter specific buyers, and get top spenders:
```

```python
Total revenue per category
```

```python
category_revenue = {}
for name, product, price, category in orders:
    category_revenue[category] = category_revenue.get(category, 0.0) + price
```

```python
Extract all unique products
```

```python

unique_products = {product for name, product, price, category in orders}
```

```python
list comprehension for customers who bought Electronics
```

```python

electronics_buyers = list({name for name, product, price, category in orders if category == "Electronics"})
```

```python
Top 3 highest-spending customers
```

```python
top_3_spenders = sorted(customer_totals.items(), key=lambda x: x[1], reverse=True)[:3]

print("Category Revenue:", category_revenue)
print("Unique Products:", unique_products)
print("Electronics Buyers:", electronics_buyers)
print("Top 3 Spenders:", top_3_spenders)
```

```python
                                ****Task 5: Organize & Display Summary with Set Operations****
Set Comparisons
```

```python
Print customer summaries
```

```python
print("--- CUSTOMER SUMMARY ---")
for name in customer_totals:
    print(f"Customer: {name} | Total Spent: ${customer_totals[name]:.2f} | Category: {customer_classes[name]}")
```

```python
Customers who bought from specific categories using sets
```

```python

electronics_set = {name for name, prod, price, cat in orders if cat == "Electronics"}
clothing_set = {name for name, prod, price, cat in orders if cat == "Clothing"}
```

```python
Intersections and Multi-category buyers
```

```python

both_elec_and_cloth = electronics_set.intersection(clothing_set)

customer_categories = {}
for name, prod, price, cat in orders:
    customer_categories.setdefault(name, set()).add(cat)

multi_cat_buyers = {name for name, cats in customer_categories.items() if len(cats) > 1}

print("\n--- INSIGHTS ---")
print("Bought both Electronics & Clothing:", both_elec_and_cloth)
print("Purchased from Multiple Categories:", multi_cat_buyers)
```
